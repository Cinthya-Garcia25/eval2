```python
from pathlib import Path

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.model_selection import GridSearchCV, KFold, train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, classification_report, roc_curve, roc_auc_score,
)

ruta = Path("..")
datos = pd.read_csv(ruta / "data" / "breast_cancer.csv")
datos = datos.drop(columns=[c for c in datos.columns if c.startswith("Unnamed") or c == "id"])
datos["diagnosis"] = datos["diagnosis"].map({"M": 1, "B": 0})
datos.head()
```

```python
print("Dimensiones:", datos.shape)
print("Valores nulos:", datos.isna().sum().sum())
print("Balance de clases (1=Maligno, 0=Benigno):")
print(datos["diagnosis"].value_counts())
```

```python
BLUE = "#2a78d6"
ORANGE = "#eb6834"
MUTED = "#898781"
GRID = "#e1e0d9"
SURFACE = "#fcfcfb"

plt.rcParams.update({
    "figure.facecolor": SURFACE, "axes.facecolor": SURFACE,
    "axes.edgecolor": GRID, "grid.color": GRID, "grid.linewidth": 0.8,
    "axes.spines.top": False, "axes.spines.right": False,
})

conteo = datos["diagnosis"].value_counts().sort_index()
fig, ax = plt.subplots(figsize=(6, 4.5), dpi=120)
ax.bar(["Benigno (0)", "Maligno (1)"], conteo.values, color=[BLUE, ORANGE], width=0.55)
ax.set_ylabel("Numero de casos")
ax.set_title("Balance de clases — breast_cancer.csv", loc="left")
for i, v in enumerate(conteo.values):
    ax.text(i, v + 5, str(v), ha="center", color="#0b0b0b")
fig.tight_layout()
fig.savefig(ruta / "figures" / "sa_clasificacion_balance.png", facecolor=SURFACE)
plt.show()
```

```python
x = datos.drop(columns=["diagnosis"])
y = datos["diagnosis"]

x_entrena, x_prueba, y_entrena, y_prueba = train_test_split(
    x, y, test_size=0.2, random_state=42, stratify=y
)

validacion = KFold(n_splits=5, shuffle=True, random_state=42)
print(f"Entrenamiento: {x_entrena.shape[0]} casos | Prueba: {x_prueba.shape[0]} casos")
```

```python
modelo_base = DecisionTreeClassifier(random_state=42)
modelo_base.fit(x_entrena, y_entrena)
pred_base = modelo_base.predict(x_prueba)
print("Accuracy modelo base (sin optimizar):", round(accuracy_score(y_prueba, pred_base), 3))
print("Profundidad del arbol base:", modelo_base.get_depth())
```

```python
rejilla = {
    "max_depth": [2, 3, 4, 5, 6, 8, None],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4],
    "criterion": ["gini", "entropy"],
}

busqueda = GridSearchCV(
    DecisionTreeClassifier(random_state=42), rejilla, cv=validacion,
    scoring="f1", n_jobs=-1
)
busqueda.fit(x_entrena, y_entrena)

print("Mejores hiperparametros:", busqueda.best_params_)
print("Mejor F1 (CV, train):", round(busqueda.best_score_, 3))

modelo_final = busqueda.best_estimator_
pred_final = modelo_final.predict(x_prueba)
prob_final = modelo_final.predict_proba(x_prueba)[:, 1]
```

```python
metricas = {
    "Accuracy": accuracy_score(y_prueba, pred_final),
    "Precision": precision_score(y_prueba, pred_final),
    "Recall": recall_score(y_prueba, pred_final),
    "F1": f1_score(y_prueba, pred_final),
    "ROC-AUC": roc_auc_score(y_prueba, prob_final),
}
resultados = pd.DataFrame([metricas]).round(3)
print(classification_report(y_prueba, pred_final, target_names=["Benigno", "Maligno"]))
resultados
```

```python
cm = confusion_matrix(y_prueba, pred_final)
fig, ax = plt.subplots(figsize=(5.5, 5), dpi=120)
im = ax.imshow(cm, cmap="Blues")
etiquetas = ["Benigno", "Maligno"]
ax.set_xticks([0, 1]); ax.set_xticklabels(etiquetas)
ax.set_yticks([0, 1]); ax.set_yticklabels(etiquetas)
ax.set_xlabel("Prediccion"); ax.set_ylabel("Real")
ax.set_title("Matriz de confusion — Arbol de Decision optimizado", loc="left")
for i in range(2):
    for j in range(2):
        color = "white" if cm[i, j] > cm.max() / 2 else "#0b0b0b"
        ax.text(j, i, str(cm[i, j]), ha="center", va="center", color=color, fontsize=14)
fig.tight_layout()
fig.savefig(ruta / "figures" / "sa_matriz_confusion.png", facecolor=SURFACE)
plt.show()
```

```python
fpr, tpr, _ = roc_curve(y_prueba, prob_final)
auc = roc_auc_score(y_prueba, prob_final)

fig, ax = plt.subplots(figsize=(6, 5.5), dpi=120)
ax.plot(fpr, tpr, color=ORANGE, linewidth=2, label=f"Arbol de Decision (AUC = {auc:.3f})")
ax.plot([0, 1], [0, 1], color=MUTED, linestyle="--", linewidth=1.2, label="Clasificador aleatorio")
ax.set_xlabel("Tasa de falsos positivos")
ax.set_ylabel("Tasa de verdaderos positivos")
ax.set_title("Curva ROC", loc="left")
ax.legend(frameon=False, loc="lower right", fontsize=9)
fig.tight_layout()
fig.savefig(ruta / "figures" / "sa_clasificacion_roc.png", facecolor=SURFACE)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(22, 12), dpi=110)
plot_tree(
    modelo_final, feature_names=x.columns, class_names=["Benigno", "Maligno"],
    filled=True, rounded=True, fontsize=8, ax=ax
)
ax.set_title(f"Arbol de Decision completo (max_depth={busqueda.best_params_['max_depth']})", fontsize=14)
fig.tight_layout()
fig.savefig(ruta / "figures" / "sa_arbol_completo.png", facecolor=SURFACE)
plt.show()
```

```python
joblib.dump(modelo_final, ruta / "models" / "sa_arbol_breast_cancer.joblib")
print("Modelo guardado en la carpeta models/")
```
